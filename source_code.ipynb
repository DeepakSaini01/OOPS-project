```python
from transformers import pipeline
```

```python
import os
import hashlib
import warnings

os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"
os.environ["TRANSFORMERS_VERBOSITY"] = "error"

warnings.filterwarnings("ignore")


from transformers.utils import logging

logging.set_verbosity_error()
logging.disable_progress_bar()


class NLPApp:

    SENTIMENT_MODEL = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"

    def __init__(self):
        self.database = {}
        self.sentiment_model = None
        self.__first_menu()

    @staticmethod
    def __hash(password):
        return hashlib.sha256(password.encode()).hexdigest()

    def __first_menu(self):
        while True:
            choice = input("""
Hi! How would you like to proceed?

1. Not a member? Register
2. Already a member? Login
3. Exit

""").strip()

            if choice == "1":
                self.__register()

            elif choice == "2":
                if self.__login():
                    self.__second_menu()

            elif choice == "3":
                return

            else:
                print("Invalid input")

    def __second_menu(self):
        while True:
            choice = input("""
Hi! How would you like to proceed?

1. NER
2. Language Detection
3. Sentiment Analysis
4. Logout

""").strip()

            if choice == "1":
                self.__ner()

            elif choice == "2":
                self.__language_detection()

            elif choice == "3":
                self.__sentiment_analysis()

            elif choice == "4":
                return

            else:
                print("Invalid input")

    def __register(self):
        name = input("Enter name: ")
        email = input("Enter email: ")
        password = input("Enter password: ")

        if email in self.database:
            print("User already exists")
        else:
            self.database[email] = [name, self.__hash(password)]
            print("User registered successfully")

    def __login(self):
        email = input("Enter email: ")
        password = input("Enter password: ")

        if email not in self.database:
            print("User does not exist")
            return False

        if self.__hash(password) == self.database[email][1]:
            print("Login successful")
            return True

        print("Login failed")
        return False

    def __sentiment_analysis(self):
        para = input("Enter the paragraph: ")

        if self.sentiment_model is None:
            self.sentiment_model = pipeline(
                "sentiment-analysis",
                model=self.SENTIMENT_MODEL
            )

        result = self.sentiment_model(para)[0]

        print("Sentiment:", result["label"])
        print("Confidence:", round(result["score"], 4))

    def __ner(self):
        print("NER feature")

    def __language_detection(self):
        print("Language Detection feature")


if __name__ == "__main__":
    NLPApp()
```
